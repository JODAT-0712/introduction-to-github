Features separation


```python
# separating TARGET and unimportant predictive columns in the dataset.
x_train = train.drop(columns = ['TARGET','SK_ID_CURR'])
y_train = train['TARGET']
x_test = test.drop(columns = ['SK_ID_CURR'])
print("New train set: ",x_train.shape)
print("New test set: ",x_test.shape)
```

Identify numerical and categorical columns

```python
# Identify numerical and categorical columns
train_numerical = x_train.select_dtypes(include=["number"]).columns.tolist()
train_categorical = x_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

test_numerical = x_test.select_dtypes(include=["number"]).columns.tolist()
test_categorical = x_test.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("TRAIN DATASET")
print("Numerical columns:", len(train_numerical))
print("Numerical features:", train_numerical)

print("\nCategorical columns:", len(train_categorical))
print("Categorical features:", train_categorical)

print("Total columns:", x_train.shape[1])

print("\nTEST DATASET")
print("Numerical columns:", len(test_numerical))
print("Numerical features:", test_numerical)

print("\nCategorical columns:", len(test_categorical))
print("Categorical features:", test_categorical)

print("Total columns:", x_test.shape[1])
```

handling missing values in the datasets by using imputer method

```python
from sklearn.impute import SimpleImputer

# Create imputers (set add_indicator=False so output shape matches input shape)
num_imputer = SimpleImputer(strategy="median", add_indicator=False)
cat_imputer = SimpleImputer(strategy="constant", fill_value="Missing")

# Impute numerical features
x_train[train_numerical] = num_imputer.fit_transform(
    x_train[train_numerical]
)
x_test[test_numerical] = num_imputer.transform(
    x_test[test_numerical]
)

# Impute categorical features
if train_categorical:
    x_train[train_categorical] = cat_imputer.fit_transform(
        x_train[train_categorical]
    )
    x_test[test_categorical] = cat_imputer.transform(
        x_test[test_categorical]
    )

# Check remaining missing values
print("Missing values in x_train:", x_train.isna().sum().sum())
print("Missing values in x_test:", x_test.isna().sum().sum())
```

feature engineering and data preprocessing

```python
def feature_engineering(data):
    data = data.copy()

    #  Credit amount compared with income
    if {"AMT_CREDIT", "AMT_INCOME_TOTAL"}.issubset(data.columns):
        data["CREDIT_INCOME_RATIO"] = (
            data["AMT_CREDIT"] /
            data["AMT_INCOME_TOTAL"].replace(0, np.nan)
        )

    #  Annual loan payment compared with income
    if {"AMT_ANNUITY", "AMT_INCOME_TOTAL"}.issubset(data.columns):
        data["ANNUITY_INCOME_RATIO"] = (
            data["AMT_ANNUITY"] /
            data["AMT_INCOME_TOTAL"].replace(0, np.nan)
        )

    #  Credit amount compared with annual loan payment
    if {"AMT_CREDIT", "AMT_ANNUITY"}.issubset(data.columns):
        data["CREDIT_ANNUITY_RATIO"] = (
            data["AMT_CREDIT"] /
            data["AMT_ANNUITY"].replace(0, np.nan)
        )

    # Convert age from days to years
    if "DAYS_BIRTH" in data.columns:
        data["AGE_YEARS"] = data["DAYS_BIRTH"].abs() / 365.25

    #  Average available external credit scores
    ext_cols = [
        col for col in [
            "EXT_SOURCE_1",
            "EXT_SOURCE_2",
            "EXT_SOURCE_3"
        ]
        if col in data.columns
    ]

    if ext_cols:
        data["EXT_SOURCE_MEAN"] = data[ext_cols].mean(axis=1)

    #  Replace infinite values with missing values
    data.replace([np.inf, -np.inf], np.nan, inplace=True)

    return data


# Apply feature engineering to both datasets
x_train = feature_engineering(x_train)
x_test = feature_engineering(x_test)

new_features = [
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_ANNUITY_RATIO",
    "AGE_YEARS",
    "EXT_SOURCE_MEAN"
]

new_features = [
    col for col in new_features
    if col in x_train.columns
]

print("Training dataset shape:", x_train.shape)
print("Testing dataset shape:", x_test.shape)
print("Training dataset missing values :", x_train.isna().sum().sum())
print("Testing dataset missing values :", x_test.isna().sum().sum())


print("\nNew features created:")
print(new_features)

print("\nStatistics of new training features:")
display(x_train[new_features].describe())
```

```python
# CatBoost algorithm
from catboost import CatBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

# Perform a clean and consistent split to ensure matching lengths
X_train_part, X_valid, y_train_part, y_valid = train_test_split(
    x_train, y_train, test_size=0.20, stratify=y_train, random_state=42
)

model = CatBoostClassifier(
    iterations=800,
    depth=10,
    learning_rate=0.01,
    l2_leaf_reg=5,
    random_strength=2,
    loss_function="Logloss",
    eval_metric="AUC",
    auto_class_weights="Balanced",
    random_seed=42,
    verbose=100
)

# Train using correctly aligned split variables
model.fit(
    X_train_part,
    y_train_part,
    cat_features=train_categorical,
    eval_set=(X_valid, y_valid)
)

# Predictions
train_probs = model.predict_proba(X_train_part)[:, 1]
valid_probs = model.predict_proba(X_valid)[:, 1]

# ROC-AUC
print("Training ROC-AUC:", roc_auc_score(y_train_part, train_probs))
print("Validation ROC-AUC:", roc_auc_score(y_valid, valid_probs))
```
